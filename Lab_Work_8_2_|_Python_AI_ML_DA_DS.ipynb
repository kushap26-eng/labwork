#### **1. Mathematical Operations**

##### 1. Create two NumPy arrays of shape (3, 3) and perform element-wise addition, subtraction, multiplication, and division.

##### 2. Generate a NumPy array with random integers and apply square root, exponential, and logarithmic functions.

```python
import numpy as np
array = np.random.randint(1,50,5)
sqrt_arr = np.sqrt(array)
exp_arr = np.exp(array)
log_arr = np.log(array)
print("Original Array: \n",array)
print("\nSquare Root: \n",np.round(sqrt_arr,4))
print("\nExponential: \n",np.round(exp_arr,4))
print("\nNatural Logarithm: \n",np.round(log_arr,4))
```

##### 3. Create a NumPy array and apply trigonometric operations like sin, cos, and tan.

```python
import numpy as np
angles_degree = np.array([0,30,45,60,90])
print("Original Angles(Degrees): ",angles_degree)

angles_radians = np.radians(angles_degree)
print("Angles(Radians): ",angles_radians)

sine_values = np.sin(angles_radians)
cos_values = np.cos(angles_radians)
tangent_values = np.tan(angles_radians)

print("Sine: ",sine_values)
print("Cos:",cos_values)
print("Tangent:",tangent_values)
```

##### 4. Normalize a NumPy array using vectorized operations (subtract mean and divide by std).

```python
import numpy as np
array = np.array([10,20,30,40,50])
mean = np.mean(array)
std = np.std(array)
normalised = (array - mean) / std
print("Original Array:",array)
print("Mean:",mean)
print("Standard Deviation: ",std)
print("Normalised Array: ",normalised)
```

##### 5. Implement a function using NumPy that calculates the Euclidean distance between two points.

```python
import numpy as np
def euclidean_distance(point1,point2):
  pont1 = np.array(point1)
  point2 = np.array(point2)
  distance = np.sqrt(np.sum((point1-point2)**2))

  return distance

p1 = list(map(float,input("Enter coordination of first point: ").split()))
p2 = list(map(float,input("Enter coordination of second point: ").split()))

out  = euclidean_distance(p1,p2)
print("Euclidean Distance: ",out)
```

#### **2. Combining and Splitting Arrays**

##### 1. Create two 1D arrays and concatenate them horizontally and vertically.

```python
import numpy as np

arr1 = np.random.randint(1,10,3)
arr2 = np.random.randint(1,10,3)

horizontal = np.concatenate((arr1,arr2))
vertical = np.concatenate((arr1,arr2))

print("Array1: ",arr1)
print("Array2: ",arr2)

print("Horizontal Concatenation: ",horizontal)
print("Vertical Concatenation: ",vertical)
```

##### 2. Stack multiple 2D arrays using vstack and hstack and then split them back using vsplit and hsplit.

```python
import numpy as np
arr1 = np.random.randint(1,10,size=(2,2))
arr2 =np.random.randint(1,10,size=(2,2))
print("Array1:",arr1)
print("Array2:",arr2)

v_stake = np.vstack((arr1,arr2))
print("\nAfter vstake")
print(v_stake)

v_split = np.vsplit(v_stake,2)
print("\nAfter vsplit:")
for arr in v_split:
  print(arr)

h_stake= np.hstack((arr1,arr2))
print("\nAfter hstake")
print(h_stake)

h_split = np.hsplit(h_stake,2)
print("\nAfter hsplit:")
for arr in h_split:
  print(arr)
```

##### 3. Create a 2D array and split it into 4 equal sub-arrays.

```python
import numpy as np

arr = np.arange(1,17).reshape(4,4)

print("Original Array: ")
print(arr)

result = np.vsplit(arr,4)
print("4 Equal Sub-arrays:")
for a in result:
  print(a)
```

##### 4. Demonstrate the difference between concatenate, stack, and hstack functions using 2D arrays.

```python
import numpy as np
arr1 = np.random.randint(1,10,size = (2,2))
arr2 = np.random.randint(1,10,size = (2,2))

print("Array1:")
print(arr1)

print("\nArray2: ")
print(arr2)

conc = np.concatenate((arr1,arr2),axis =0)
print("\nConcatenate: ")
print(conc)

st = np.stack((arr1,arr2),axis=0)
print("\nStack:")
print(st)

hs = np.hstack((arr1,arr2))
print("\nHstack:")
print(hs)
```

#### **3. Search, Sort, Aggregating/Statistical Functions**

##### 1. Create a NumPy array and find the index of the maximum and minimum elements.

```python
import numpy as np
arr = np.array([10,25,5,40,15])

print("Array:",arr)

max_index = np.argmax(arr)
min_index = np.argmin(arr)

print("Index of maximum element: ",max_index)
print("index of minimum element: ",min_index)
```

##### 2. Sort a 2D array along different axes and display the result.

```python
import numpy as np

arr = np.random.randint(1,10,size=(3,3))

print("Original array:")
print(arr)

row_sort = np.sort(arr,axis=1)

print("\nSorted along rows:")
print(row_sort)

column_sort = np.sort(arr,axis=0)
print("\nSorted along columns:")
print(column_sort)
```

##### 4. Use np.where to find all elements greater than a certain threshold in a given array.
