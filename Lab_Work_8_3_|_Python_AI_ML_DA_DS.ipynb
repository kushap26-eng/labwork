````python
#Task 1.2
import numpy as np

ages = np.array([25,30,18,35,22,40,17])

young_adults = ages[(ages > 20) & (ages < 35)]
print(ages)
print("Young adults: ",young_adults)
````

````python
#Task1.3
import numpy as np

names = np.array(['Alice','Bob','Charlie','David','Eve'])
mask = np.char.startswith(names,'A') | np.char.startswith(names,'E')
a_or_E = names[mask]

print("Original names: ",names)
print("Names starting with 'A' or 'E': ",a_or_E)
````

````python
#Task 2.1
import numpy as np
temerature = np.array([28,32,25,35,30,22,38])
is_hot = temerature > 30
hot_temparatures = temerature[is_hot]
print("Original temparature: ",temerature)
print("Boolean mask: ",is_hot)
print("Hot temparature:   ",hot_temparatures)
````

````python
#Task2.2
import numpy as np

temerature = np.array([28,32,25,35,30,22,38])
is_hot = temerature > 30
temerature[is_hot] = 31
print("Modified temparature: ",temerature)
````

````python
#Task 2.3
import numpy as np

scores = np.array([65,80,72,90,55,78])
grades = np.where(scores >= 75, "Pass","Fail")
print("Scores: ",scores)
print("Grades: ",grades)
````
